````python

def collapse_backslashes(text: str) -> str:
    """
    Заменяет \\  \\\\  \\\\\\ … на одинарный \.
    Работает безопасно для всего текста: формул, JSON-фрагментов и т.д.
    """
    return re.sub(r'\\{2,}', r'\\', text)
````

````python
# %% 🛠️ Быстрая проверка Markdown + LaTeX
import markdown, re
from pylatexenc.latexwalker import LatexWalker, LatexWalkerParseError

def quick_md_check(text: str) -> list[str]:
    """Возвращает список найденных проблем."""
    probs = []
    txt = collapse_backslashes(text)

    # непарные ```  или $
    if txt.count("```") % 2: probs.append("непарное число ```")
    if txt.count("$")   % 2: probs.append("непарное число $")

    # заголовки без пробела
    if re.search(r"^(#{1,6})(\S)", txt, flags=re.MULTILINE):
        probs.append("нет пробела после # в заголовке")

    # валидируем LaTeX-фрагменты
    for formula in re.findall(r"\$(.*?)\$", txt):
        try:
            LatexWalker(formula).get_latex_nodes()
        except LatexWalkerParseError as e:
            probs.append(f"LaTeX: {e}")

    # markdown → html (ловим SyntaxError от парсера)
    try:
        markdown.markdown(txt, extensions=["fenced_code", "tables"])
    except Exception as e:
        probs.append(f"Markdown-parser: {e}")
    return probs
````

````python
# %% 🏃‍♂️ Главная функция для списка задач
from termcolor import colored
from tqdm import tqdm

def validate_tasks(tasks, quest_key="question"):
    bad = 0
    for i, t in enumerate(tqdm(tasks, desc="Проверка")):
        md = t[quest_key]
        errs = quick_md_check(md)
        if errs:
            bad += 1
            print(colored(f"[{i}] ❌ {errs}", "red"))
        else:
            print(colored(f"[{i}] ✅ OK", "green"))
    print(colored(f"\nИТОГО: ошибок {bad} из {len(tasks)}", "cyan"))
````
