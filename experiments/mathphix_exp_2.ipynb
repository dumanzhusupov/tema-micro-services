## Стартуем микросервис OCR на FastAPI с Mathpix

**Почему FastAPI:**  
- Автоматическая генерация Swagger UI для тестирования API.
- Асинхронная работа — удобно для интеграции с внешними сервисами (Mathpix, S3, MongoDB).
- Простая валидация входных данных через Pydantic.
- Легко масштабировать и интегрировать в микросервисную архитектуру.

**План:**  
1. Создать FastAPI-приложение.
2. Реализовать эндпоинт `/ocr/pdf` для загрузки PDF-файла.
3. Внутри эндпоинта — отправлять файл в Mathpix API, ждать результат.
4. Сохранять результат (например, в S3 или локально).
5. Возвращать ссылку на результат и метаданные через API.
6. Все шаги снабжать подробными комментариями.

**Далее — пример кода FastAPI OCR сервиса.**

```python
# main.py — пример FastAPI OCR сервиса с Mathpix
# Подробные комментарии объясняют каждое действие

from fastapi import FastAPI, UploadFile, File, HTTPException
from fastapi.responses import JSONResponse
import os
from dotenv import load_dotenv
from mpxpy.mathpix_client import MathpixClient

# Загружаем переменные окружения (API ключи и т.д.)
load_dotenv()

app = FastAPI(
    title="Mathpix OCR API",
    description="Микросервис для распознавания PDF через Mathpix и выдачи результата",
    version="0.1.0"
)

# Инициализация клиента Mathpix с ключами из .env
client = MathpixClient(
    app_id=os.getenv('APP_ID'),
    app_key=os.getenv('APP_KEY')
)

@app.post("/ocr/pdf")
async def ocr_pdf(file: UploadFile = File(...)):
    """
    Эндпоинт для загрузки PDF-файла и распознавания его через Mathpix.
    Возвращает ссылку на результат и метаданные.
    """
    # Проверяем тип файла
    if not file.filename.lower().endswith(".pdf"):
        raise HTTPException(status_code=400, detail="Требуется PDF-файл")

    # Сохраняем файл во временную директорию
    temp_path = f"/tmp/{file.filename}"
    with open(temp_path, "wb") as f:
        f.write(await file.read())

    # Отправляем PDF в Mathpix (синхронно, можно сделать асинхронно через Celery)
    pdf = client.pdf_new(file_path=temp_path)
    pdf.wait_until_complete(timeout=450)

    # Сохраняем результат в локальный файл (или можно в S3)
    result_md_path = temp_path.replace(".pdf", ".md")
    pdf.to_md_file(path=result_md_path)

    # Можно добавить сохранение в S3 и возврат ссылки
    # Здесь возвращаем только путь и статус
    return JSONResponse({
        "status": "ok",
        "result_md_path": result_md_path,
        "pages": pdf.num_pages
    })

# Для запуска: uvicorn main:app --reload
```
